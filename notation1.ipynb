```python
import numpy as np
import matplotlib.pyplot as plt

#création liste de toutes les matières du semestre 1
matiere=[f"R{i}" for i in range(101, 116)]+[f"SAE{i}" for i in range(11, 17)]

#liste des coef des trois UE
list_coef1 = [10,10,7,7,0,5,0,6,0,5,4,2,5,5,0,20,20,0,0,0,7]
list_coef2 = [4,0,2,8,6,0,0,0,0,5,5,2,9,9,3,0,0,29,0,0,7]
list_coef3 = [4,0,2,0,0,5,15,6,4,5,5,2,0,0,3,0,0,0,20,20,7]
list_notes = []

#list_notes = [float(input(f"Quelle est la note pour {element} ?")) for element in matiere]
list_notes = [i for i in range(21)] #test

def calcul_moy(list_coef):
    return round(np.average(list_notes, weights = list_coef),3) #ou moy =
print(f"Les moyennes sont {calcul_moy(list_coef1)} pour UE1, {calcul_moy(list_coef2)} pour UE2, et {calcul_moy(list_coef3)} pour l'UE3")

def couleur(moy):
    if moy >= 10:
        return "green"
    elif 8 <= moy < 10 :
        return "orange"
    else :
        return "red"

couleurs= [couleur(calcul_moy(list_coef1)), couleur(calcul_moy(list_coef2)), couleur(calcul_moy(list_coef3))]

plt.bar(["UE1","UE2","UE3"], [calcul_moy(list_coef1), calcul_moy(list_coef2), calcul_moy(list_coef3)], color=["r","g","g"])
```
